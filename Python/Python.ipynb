# Automatic File Sorter in File Explorer


```python
import os, shutil
```

```python
path = r"C:/Users/omkar/OneDrive/Desktop/Data Analyst/Python/Demo/"
```

```python
file_name = os.listdir(path)
```

```python
folder_names = ['csv files','txt files','images']

for file in range(0,3):
    if not os.path.exists(path + folder_names[file]):
        # print(path + folder_names[file])
        os.makedirs(path + folder_names[file])

for file in file_name:
    if '.csv' in file and not os.path.exists(path + "csv files/" + file):
        shutil.move(path + file, path +  "csv files/" + file)
    elif '.txt' in file and not os.path.exists(path + "txt files/" + file):
        shutil.move(path + file, path +  "txt files/" + file)
    elif '.png' in file and not os.path.exists(path + "images/" + file):
        shutil.move(path + file, path +  "images/" + file)
```

## Classes & Objects

```python
class Human :
    # Self Method - Generated Automatically or can be written
    def __init__(self,name,occupation,age):
        self.name = name # Properties
        self.occupation = occupation
        self.age = age 

    # Methods1 - Any type of Work
    def work(self) :
        if self.occupation == "Cricket":
            print(self.name, "play Sports")
        elif self.occupation == "Service":
            print(self.name, "Work in IT")
        elif self.occupation == "Household":
            print(self.name, "Do Cooking")
        else :
            print(self.name, "Do Study")

    # Method2
    def speak(self) :
        print(self.name, "How are You ?")


# Creating Instance of Class or Object Creation
N1 = Human("A","Service",25)
N1.work()
N1.speak()

# Instance 2
N2 = Human("B","Cricket",25)
N2.work()
N2.speak()

N3 = Human("C","Household",25)
N3.work()
N3.speak()

N4 = Human("D","",25)
N4.work()
N4.speak()
```

## Inheritance

```python
class Vehicle :
    def general_use(self) :
        print("Parent Class")

class Car(Vehicle) :
    def __init__(self) :
        print("Car Class")
        self.general_use()
        self.wheels = 4
        self.roof = True

    def usage(self) :
        print("Car use : Travel Long")

class Bike(Vehicle) :
    def __init__(self) :
        print("Bike Class")
        self.general_use()
        self.wheels = 2
        self.roof = False

    def usage(self):
        print("Bike use : Travel Short")


C = Car()
C.usage()

M = Bike()
M.usage()


# Isinstance 
print(isinstance(C,Car))
print(isinstance(C,Bike))
print(isinstance(M,Bike))
print(isinstance(M,Car))

# Issubclass Methods
print(issubclass(Car,Vehicle))
print(issubclass(Car,Bike))
print(issubclass(Bike,Vehicle))
print(issubclass(Bike,Car))
```

## Multiple Inheritance

```python
class Father():
    def Income(self):
        print("Earn Money")

class Mother():
    def Management(self):
        print("Manage Money")

class Child(Father,Mother):
    def Enjoy(self):
        print("Enjoy with Money")

c = Child()
c.Income()
c.Management()
c.Enjoy()
```

```python
class Father():
    def Skills(self):
        print("Earn Money,Make Money")

class Mother():
    def Skills(self):
        print("Manage Money,Invest Money")

class Child(Father,Mother):
    def Skills(self):
        Father.Skills(self)
        Mother.Skills(self)
        print("Play Games,Enjoy with Money")

c = Child()
c.Skills()
```

## Generators

```python
# Return Values one by one - Best for large Data Handling, Preserves Previous Result 
def gen() :
    yield "A"
    yield "B"
    yield "C"
    yield "D"

a = gen()
a
next(a)
next(a)
```

```python
# Fibnocci Sequence with Generator
def fibonacci():
    a,b = 0,1
    while True :
        yield a
        a,b = b, a+b

for f in fibonacci():
    if f < 50:
        break
    print(f, end = " ")
```
