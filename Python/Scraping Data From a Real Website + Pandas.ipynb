# Scraping Data From a Real Website + Pandas 2

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://companiesmarketcap.com/inr/'

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
soup
```

```python
soup.find_all('table')
```

```python
soup.find('table', class_ ='default-table table marketcap-table dataTable' )
```

```python
table = soup.find('table', class_ ='default-table table marketcap-table dataTable' )
```

```python
table
```

```python
titles = table.find_all('th')
```

```python
titles
```

```python
table_titles = [title.text.strip() for title in titles[1:]]

print(table_titles)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = table_titles)
```

```python
df
```

```python
column_data = table.find_all('tr')

for row in column_data:
    row_data = row.find_all('td')
    
    # [1:] skips the first 'td' (the favorite column)
    individual_row_data = [data.text.strip() for data in row_data[1:]] 
    
    if len(individual_row_data) == len(df.columns):
        length = len(df)
        df.loc[length] = individual_row_data
        
```
