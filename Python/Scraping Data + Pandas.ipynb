# Scraping Data From a Real Website + Pandas 

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
print(soup)
```

```python
soup.find('table')
```

```python
soup.find_all('table')[1]
```

```python
soup.find('table', class_ ="wikitable sortable jquery-tablesorter")
```

```python
table = soup.find_all('table', class_ ='wikitable sortable jquery-tablesorter')
```

```python
print(table)
```

```python
world_titles = table.find_all('th')
```

```python
world_titles
```

```python
world_table_titles = [title.text.strip() for title in world_titles]

print(world_table_titles)
```

```python
import pandas as pd
```

```python
df = pd.Dataframes(columns = world_table_titles)
```

```python
df
```

```python
column_data = table.find_all('tr')

for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]

    length = len(df)
    df.loc[length] = individual_row_data
    
```

```python
df
```

```python
df.to_scv(r'path',index = False)
```
