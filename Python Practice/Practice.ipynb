```python
# Even or Odd No
n = int(input("Enter No"))
if n==0:
    print("Not Even no Odd")
elif n%2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Prime No
a = int(input("Enter No"))
if (a==0 | a==1):
    print("No Prime No Composite")
elif a%2 == 0:
    print('Prime')
else:
    print('Composite')
```

```python
# Fibonacci Series 
n = 20
def fibonacci(n):
    a,b = 0,1
    list1 = []
    for i in range(n):
        list1.append(a)
        a,b = b,a+b
    return list1
```

```python
print(fibonacci(30))
```

```python
# Factorial by Recursive Method
n = 10
def factorial(n):
    if n == 0 | n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(10))
```

```python
# Factorial by Normal Method
def factorial_iterative(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

print(factorial_iterative(10))
```

```python
# Factorial by Formula Method
import math
print(math.factorial(10))
```
