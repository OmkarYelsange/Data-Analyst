# Probablity
- P = No of Favorable Outcome/Total Outcome
- Always lies between 0 to 1

## Addition Rule of Probability
1. Two Events Mutually Exculsive
- P(A U B) = P(A) + P(B)
2. Two Events Mutually Non-Exculsive
- P(A U B) = P(A) + P(B) - P(A n B)

## Independent Events
- P(A/B) = P(A)
- P(B/A) = P(B)
- P(A&B) = P(A) * P(B)

## Cumulative Probability
- Probability that a value of the given Random Variable is within a given range
- Eg. P(a <= X <= b)
