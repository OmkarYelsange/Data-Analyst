# Statistics
### It is a Branch of Mathematics which is used for analysis of large data using Charts graphs and etc 

## Types of Stats
1. Descriptive Statistics
2. Inferential Statistics
3. Types of Data

## Descriptive Statistics
### A Set of Methods used Describes the Characteristics of a Data Set. 
1. Mode - Average
2. Median - Middle value (sorted data)
3. Mode - Max no of Occurance
4. Variance - (Sum) sq / N
5. SD = Sqrt of Variance
6. Count 

## Inferential Statistics

### A Set of Methods used to draw Conclusion about the Characteristics of Population based on Sample Data.

1. Hypothesis Testing
2. Estimating Parameter

## Types of Data 

### 1. Qualitative / Categorical Data - Norminal, Ordinal Data.
Eg - Gender, Hometown, Birthdate, Postcode, etc
### Norminal Data
- Cannot be Ordered or Measured.
- Eg. Gender, Hometown
### Ordinal Data
- Can be Ordered
- Eg. Grade, Economics, Finance
